```python
import os
import random
import shutil
import pandas as pd
import re
```

```python
# Directorios de entrenamiento y validación
train_dir = 'Birds_Species_Dataset/train'
valid_dir = 'Birds_Species_Dataset/valid_increased'

dir_csv = 'Birds_Species_Dataset/birds.csv'
df = pd.read_csv(dir_csv)

```

```python
for class_name in  os.listdir(train_dir):
    print(class_name)
```

```python
# Número de imágenes a copiar por clase
num_images_to_copy = 10

# Recorrer cada carpeta/clase en el conjunto de entrenamiento
for class_name in os.listdir(train_dir):
    train_class_dir = os.path.join(train_dir, class_name)
    valid_class_dir = os.path.join(valid_dir, class_name)

    # Crear la carpeta en el conjunto de validación si no existe
    os.makedirs(valid_class_dir, exist_ok=True)

    # Listar todas las imágenes en la carpeta de entrenamiento
    images = os.listdir(train_class_dir)

    # Seleccionar aleatoriamente 10 imágenes
    images_to_copy = random.sample(images, num_images_to_copy)

    # Copiar cada imagen seleccionada al conjunto de validación
    for image in images_to_copy:
        src_path = os.path.join('train/', class_name, image)
        dest_path = os.path.join(valid_class_dir, image)
        src_path = src_path.replace('\\', '/')
        dest_path = dest_path.replace('Birds_Species_Dataset/', '')
        dest_path = dest_path.replace('\\', '/')
        print(src_path)
        print('separator')
        print(dest_path)
        original_row = df[df['filepaths'] == src_path].iloc[0]
        shutil.copy(os.path.join(train_class_dir, image), valid_class_dir)
        original_row = df[df['filepaths'] == src_path].iloc[0]
        new_row = original_row.copy()
        new_row['filepaths'] = dest_path
        new_row['data set'] = 'valid'
        #df = df.append(new_row, ignore_index=True)
        df = pd.concat([df, pd.DataFrame([new_row])], ignore_index=True)
    print(f"Copied {num_images_to_copy} images from {train_class_dir} to {valid_class_dir}")
```

```python
#import pandas as pd 
dir_csv_new = 'Birds_Species_Dataset/birds_updated.csv'

df.to_csv(dir_csv_new, index=False)
```
